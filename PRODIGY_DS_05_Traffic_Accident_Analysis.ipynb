```python
import pandas as pd

df = pd.read_csv('/content/traffic_accidents.csv')

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

df.head()
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print("Dataset Shape:", df.shape)

print("\nBasic Statistics:")
display(df.describe(include='all').T)
```

```python
print("All Columns:")
for i, col in enumerate(df.columns, 1):
    print(i, "→", col)
```

```python
import matplotlib.pyplot as plt

weather_counts = df['weather_condition'].value_counts()

plt.figure(figsize=(10, 6))
weather_counts.plot(kind='bar')

plt.title('Number of Accidents by Weather Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
road_counts = df['roadway_surface_cond'].value_counts()

plt.figure(figsize=(10, 6))
road_counts.plot(kind='bar', color='orange')

plt.title('Number of Accidents by Road Surface Condition')
plt.xlabel('Road Surface Condition')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
hour_counts = df['crash_hour'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
plt.plot(hour_counts.index, hour_counts.values, marker='o')

plt.title('Number of Accidents by Hour of the Day')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Accidents')

plt.xticks(range(0, 24))
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
day_counts = df['crash_day_of_week'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
day_counts.plot(kind='bar', color='teal')

plt.title('Number of Accidents by Day of the Week')
plt.xlabel('Day of Week')
plt.ylabel('Number of Accidents')

plt.tight_layout()
plt.show()
```

```python
cause_counts = df['prim_contributory_cause'].value_counts().head(10)

plt.figure(figsize=(10, 6))
cause_counts.sort_values().plot(kind='barh', color='crimson')

plt.title('Top 10 Contributing Causes of Accidents')
plt.xlabel('Number of Accidents')
plt.ylabel('Contributing Cause')

plt.tight_layout()
plt.show()
```

```python
hotspot = df['trafficway_type'].value_counts().head(10)

plt.figure(figsize=(10, 6))
hotspot.sort_values().plot(kind='barh', color='purple')

plt.title('Top Roadway Types with Accident Records')
plt.xlabel('Number of Accidents')
plt.ylabel('Roadway Type')

plt.tight_layout()
plt.show()
```

```python
lighting_counts = df['lighting_condition'].value_counts()

plt.figure(figsize=(10, 6))
lighting_counts.plot(kind='bar', color='gold')

plt.title('Number of Accidents by Lighting Condition')
plt.xlabel('Lighting Condition')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
weather_road = pd.crosstab(
    df['weather_condition'],
    df['roadway_surface_cond']
)

plt.figure(figsize=(12, 7))
weather_road.plot(kind='bar', figsize=(12, 7))

plt.title('Accidents by Weather and Road Surface Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
injury_counts = df['most_severe_injury'].value_counts()

plt.figure(figsize=(10, 6))
injury_counts.plot(kind='bar', color='steelblue')

plt.title('Accidents by Most Severe Injury')
plt.xlabel('Most Severe Injury')
plt.ylabel('Number of Accidents')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
