```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/train.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
 # Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Drop Cabin column because it has many missing values
df.drop('Cabin', axis=1, inplace=True)

# Fill missing Embarked values with most frequent value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Check missing values again
df.isnull().sum()
```

```python
survival_counts = df['Survived'].value_counts()

print("Not Survived (0):", survival_counts[0])
print("Survived (1):", survival_counts[1])
```

```python
sns.countplot(x='Survived', data=df)

plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Age'], bins=20, kde=True)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x='Survived', y='Age', data=df)

plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Age')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['Fare'], bins=30, kde=True)

plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x='Survived', y='Fare', data=df)

plt.title('Fare Distribution by Survival')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')

plt.show()
```

## Key Findings and Observations
1. The dataset contains 891 passenger records and 12 original columns.

2. Missing values were found in Age, Cabin, and Embarked columns. Age was filled using the median, Cabin was removed due to many missing values, and missing Embarked values were filled using the mode.

3. Out of 891 passengers, 342 survived and 549 did not survive.

4. The gender analysis shows a clear difference in survival outcomes between male and female passengers.

5. Survival outcomes also varied across passenger classes.

6. The age distribution shows that passengers belonged to a wide range of age groups.

7. Fare values were unevenly distributed, with most passengers paying relatively lower fares and a smaller number paying higher fares.

8. The analysis shows differences in fare distributions between passengers who survived and those who did not.
